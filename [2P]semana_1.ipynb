### Juego del Ahorcado

En este ejercicio deberá implementar una versión simplificada del juego del ahorcado.

Cree la función `crear_palabra` que reciba una lista de palabras y retorne una palabra seleccionada aleatoriamente.

Luego, desarrolle un programa que:

- Utilice la función `crear_palabra` para seleccionar una palabra secreta.
- Muestre al usuario la palabra oculta utilizando guiones bajos (`_`), uno por cada letra.
- Permita al usuario ingresar una letra en cada intento.
- Si la letra pertenece a la palabra, se deben revelar todas sus apariciones.
- Si la letra no pertenece a la palabra, el usuario pierde un intento.
- El jugador dispone de **6 intentos** para descubrir la palabra.
- Si el usuario ingresa más de un carácter o un número, se debe mostrar el mensaje: `Entrada inválida. Ingrese una sola letra`
- Si el usuario ingresa una letra que ya había utilizado, se debe mostrar el mensaje: `Ya intentó con esa letra`
- Si el jugador descubre la palabra antes de quedarse sin intentos, se debe mostrar el mensaje: `¡Felicidades! Has descubierto la palabra: palabra`
- Si se terminan los intentos, se debe mostrar: `¡Juego terminado! La palabra era: palabra`

Utilice la siguiente lista de palabras:
`palabras= ["python","computadora","pirata","programacion","algoritmo","variable"]`

**Ejemplo de ejecución:**

```
¡Bienvenido al juego del ahorcado!

Palabra: _ _ _ _ _ _
Intentos restantes: 6
Ingrese una letra: p

¡Correcto!

Palabra: p _ _ _ _ _
Intentos restantes: 6
Ingrese una letra: x

Incorrecto.

Palabra: p _ _ _ _ _
Intentos restantes: 5
Ingrese una letra: y

¡Correcto!

Palabra: p y _ _ _ _
...
```

###
**Solución Propuesta:**

````python
import random

palabras = [
"python",
"computadora",
"pirata",
"programacion",
"algoritmo",
"variable"
]

def crear_palabra(lista_palabras):
    return random.choice(lista_palabras)

# Programa principal
palabra = crear_palabra(palabras)
oculta = ["_"] * len(palabra)

intentos = 6
letras_usadas = []

print("¡Bienvenido al juego del ahorcado!")

while intentos > 0 and "_" in oculta:
    print("\nPalabra:", " ".join(oculta))
    print("Intentos restantes:", intentos)

    letra = input("Ingrese una letra: ").lower()

    if len(letra) != 1 or not letra.isalpha():
        print("Entrada inválida. Ingrese una sola letra.")

    elif letra in letras_usadas:
        print("Ya intentó con esa letra.")

    else:
        letras_usadas.append(letra)

        if letra in palabra:
            print("\n¡Correcto!")

            for i in range(len(palabra)):
                if palabra[i] == letra:
                    oculta[i] = letra
        else:
            print("\nIncorrecto.")
            intentos-=1

if "_" not in oculta:
    print("\n¡Felicidades! Has descubierto la palabra:", palabra)
else:
    print("\n¡Juego terminado! La palabra era:", palabra)
````

## Ejercicio 2
Implemente un programa en Python que permita registrar los puntajes de los jugadores de una partida en un juego cooperativo.

El programa recibirá entradas en el formato:
`Nombre,puntaje`

**Ejemplo de ejecución:**

```
Registro de puntajes (nombre,puntaje). Escriba FIN para terminar.
Ingrese puntos: ana,30
Ingrese puntos: pedro,20
Ingrese puntos: luis,25
Ingrese puntos: ANA,15
Ingrese puntos: pedro,40
Ingrese puntos: FIN
```

El programa debe incluir lo siguiente:

- La captura de datos termina cuando se ingresa la palabra `FIN`.
- Cada jugador puede ingresar sus puntos varias veces (porque se juegan varias rondas).
- Debe construirse un diccionario donde:
    - La clave es el nombre del jugador (la primera letra en mayúsculas).
    - El valor es la lista de números que contiene todos los puntajes obtenidos.

```python
{'Ana': [30, 15], 'Pedro': [20, 40], 'Luis': [25]}
```


Al final se debe mostrar el puntaje total de cada jugador y el puntaje total como equipo.

**Ejemplo de ejecución:**

```
Totales
Ana: 45
Pedro: 60
Luis: 25
Total: Ganaron 130 puntos como equipo
```

###
**Solución Propuesta:**

````python
# Parte 1
print("Registro de puntajes (nombre,puntaje). Escriba FIN para terminar.")

ingreso = ''
dic = { }

while ingreso != "FIN":
    ingreso = input("Ingrese puntos: ").upper()

    if ingreso != "FIN":
        jugador, puntos = ingreso.split(",")
        puntos = int(puntos)
        jugador = jugador.title()

        if jugador not in dic:
            dic[jugador] = [puntos]
        else:
            dic[jugador].append(puntos)


# Parte 2
print('Totales')

total_equipo = 0

for jug, puntos in dic.items():
    total = sum(puntos)
    print(f"{jug}: {total}")
    total_equipo += total

print(f"Total: Ganaron {total_equipo} puntos como equipo")
````
