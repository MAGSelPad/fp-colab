```python
import pandas as pd

df = pd.DataFrame({
    "fecha": [
        "01/07/2024", "18/05/2024", "23/02/2024", "15/12/2024", "12/10/2024",
        "05/03/2024", "11/08/2024", "29/09/2024", "14/01/2024", "07/11/2024",
        "19/06/2024", "03/04/2024", "22/07/2024", "30/10/2024", "16/02/2024"
    ],
    "id_producto": [
        "B0946W4VJ9", "B0BNL81YV7", "B083LTB6CP", "B0C339DBNZ", "B0DK42VCQ4",
        "B0123ABCD1", "B0456EFGH2", "B0789IJKL3", "B0111MNOP4", "B0222QRST5",
        "B0333UVWX6", "B0444YZAB7", "B0555CDEF8", "B0666GHIJ9", "B0777KLMN0"
    ],
    "marca": [
        "Ted Baker", "Adidas", "Besttravel", "Skechers", "Casio",
        "Adidas", "Nike", "Casio", "Adidas", "Puma",
        "Nike", "Ted Baker", "Adidas", "Besttravel", "Skechers"
    ],
    "descripcion": [
        "Bolso de cuero", "Gorra deportiva", "Mochila para portátil", "Zapato de punta de acero", "Reloj informal F-91WB",
        "Camiseta deportiva", "Zapatillas Running", "Reloj deportivo", "Sudadera con capucha", "Mochila urbana",
        "Balón de fútbol", "Cartera de cuero", "Short deportivo", "Maleta de viaje", "Botas de senderismo"
    ],
    "precio": [
        109.00, 34.00, 61.99, 208.66, 115.44,
        45.50, 129.99, 89.99, 72.30, 58.90,
        39.99, 80.00, 49.95, 150.75, 180.20
    ],
    "categoria": [
        "Bolsos de mano", "Nueva temporada", "Equipaje", "Zapatos", "Relojes",
        "Ropa deportiva", "Zapatos", "Relojes", "Ropa deportiva", "Equipaje",
        "Deportes", "Accesorios", "Ropa deportiva", "Equipaje","Zapatos"
    ],
    "valoracion": [
        3.8, 4.8, 4.2, 4.6, 4.7,
        4.5, 4.9, 4.3, 4.1, 4.0,
        4.6, 3.9, 4.4, 4.8, 4.5
    ]
})


# Literal 1
masCostoso = df["precio"].idxmax()
resultado = df.loc[masCostoso, ["id_producto", "descripcion", "valoracion"]]
print(f"El producto más costoso es:\n{resultado}")


# Literal 2
filtro = df["marca"] == "Adidas"
adidas = df[filtro]

promedio = adidas["precio"].mean()
print(f"El precio promedio de los productos de marca Adidas es: {promedio}")

minimo = adidas["valoracion"].min()
print(f"La valoración más baja de los productos Adidas es: {minimo}")


# Literal 3
masBaratos = df.sort_values("precio").head(5)
masBaratos = masBaratos.sort_values("precio", ascending=False)
print(masBaratos)


# Literal 4
datos = df.groupby("categoria").agg(
    num_productos=("id_producto", "count"),
    promedio=("valoracion", "mean")
)
print(datos)
```
