````python
# PARTE 1
dic_reproducciones = {
    "Bohemian Rhapsody": ["Queen", 355, 19203],
    "We Are the Champions": ["Queen", 179, 98200],
    "We Will Rock You": ["Queen", 122, 22302],
    "Under Pressure": ["Queen", 248, 152230],
    "Shake It Off": ["Taylor Swift", 219, 145320],
    "Blank Space": ["Taylor Swift", 231, 120150],
    "Love Story": ["Taylor Swift", 235, 98420],
    "Stairway to Heaven": ["Led Zeppelin", 482, 85420],
    "Kashmir": ["Led Zeppelin", 515, 62410],
    "HUMBLE.": ["Kendrick Lamar", 177, 113520],
    "DNA.": ["Kendrick Lamar", 185, 95310],
    "N95": ["Kendrick Lamar", 196, 73210],
}

def cargar_cantante(dic_reproducciones, cantante):

    dic_cantante = {}

    for cancion, datos in dic_reproducciones.items():
        artista, duracion, reproducciones = datos

        if artista == cantante:
            dic_cantante[cancion] = reproducciones

    return dic_cantante

# PARTE 2
dic_tipos = {
    "Bohemian Rhapsody": 1,
    "We Are the Champions": 1,
    "We Will Rock You": 2,
    "Under Pressure": 1,
    "Shake It Off": 2,
    "Blank Space": 2,
    "Love Story": 2,
    "Stairway to Heaven": 1,
    "Kashmir": 1,
    "HUMBLE.": 3,
    "DNA.": 3,
    "N95": 3
}

def regalias(dic_cantante, dic_tipos):
    dic_regalias = {}
    tasas = {
        1: 0.0028,
        2: 0.0024,
        3: 0.0022
    }
    for cancion, reproducciones in dic_cantante.items():
        tipo = dic_tipos[cancion]
        monto = reproducciones * tasas[tipo]
        dic_regalias[cancion] = monto

    return dic_regalias


# PARTE 3
def total_regalias(dic_regalias):
    total = 0
    for monto in dic_regalias.values():
        total += monto
    return round(total, 2)

````
