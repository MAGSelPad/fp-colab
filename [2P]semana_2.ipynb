## Ejercicio 1

Escriba un programa para simular un juego donde el jugador debe adivinar **3 premios** antes de quedarse sin vidas.

Al inicio del programa:

1. Mostrar una lista de premios, por ejemplo:
    
    ```python
    l_premios = ['Carro', 'Moto', 'Celular', 'Smartwatch', 'Chocolate', 'Laptop', 'Caramelo']
    ```
    
2. Seleccionar **3 premios al azar sin repetir** de la lista.
3. Asignar **2 vidas** al jugador.

Durante el juego:

- En cada iteración se deben mostrar las vidas restantes.
- Solicitar al usuario que ingrese un premio.
- Si el premio fue seleccionado, mostrar el mensaje **"Adivinó"**.
- Si el premio **no** fue seleccionado, restar una vida y mostrar **"No adivinó"**.
- Puede asumir que el jugador **no ingresará el mismo premio más de una vez**.

El juego termina cuando:

- El jugador adivina los **3 premios**, o
- Se queda sin vidas.

Al finalizar, mostrar:

- **"Ganó"**, si adivinó los tres premios.
- **"Perdió"**, en caso contrario.

---

**Ejemplo de ejecución:**

```
Adivine los 3 premios seleccionados:
['Carro', 'Moto', 'Celular', 'Smartwatch', 'Chocolate', 'Laptop', 'Caramelo']

Vidas: 2. Ingrese un premio: Celular
No adivinó

Vidas: 1. Ingrese un premio: Carro
Adivinó

Vidas: 1. Ingrese un premio: Moto
Adivinó

Vidas: 1. Ingrese un premio: Smartwatch
No adivinó

Perdió
```

**Desarrolle:**

````python
# Desarrolle solución del EJERCICIO 1 aquí

````

###
**Solución Propuesta:**

````python
import random as rd

l_premios = ['Carro', 'Moto', 'Celular', 'Smartwatch', 'Chocolate', 'Laptop']

seleccionados = rd.sample(l_premios, 3)
vidas = 2

print("Adivine los 3 premios seleccionados:")
print(l_premios)
print()

while vidas > 0 and len(seleccionados) > 0:
    premio = input(f"Vidas: {vidas}. Ingrese un premio: ").title()

    if premio in seleccionados:
        print("Adivinó")
        seleccionados.remove(premio)
    else:
        print("No adivinó")
        vidas -= 1

print()

if len(seleccionados) == 0:
    print("Ganó")
else:
    print("Perdió")
````

## Ejercicio 2

Implemente la función `verificar(cedula)` que reciba un string con los 10 dígitos de un número de cédula ecuatoriana. La función debe retornar True si la cédula es válida o False en caso contrario. Para verificar la cédula, utilice el siguiente algoritmo:

1. Se agrupan los nuevos (9) primeros dígitos en dos partes: los que están en índices pares y los que están en índices impares.
2. Cada dígito en índice par se multiplica por dos (2). Solo si el resultado de esa multiplicación es mayor a nueve, se le resta nueve (9).
3. Los dígitos en índices impares se mantienen inalterados.
4. Se suman los dígitos del paso 2 y 3.
5. Se toma el dígito más a la derecha del resultado de la suma del paso 4.
6. Si el resultado obtenido en el paso 5 es cero, entonces el dígito verificador es cero. En caso contrario, el dígito verificador será el número 10 menos el dígito del paso 5.
7. Si el dígito verificador es igual al último dígito de la cédula, la cédula es válida.

**Ejemplo del proceso de validación:**

```
                    cédula => 1713175071
  Dígitos en índices pares => 1, 1, 1, 5 y 7
       Multiplicados por 2 => 2, 2, 2, 10 y 14
                Restados 9 => 2, 2, 2, 1 y 5

Dígitos en índices impares => 7, 3, 7 y 0
Suma de los nuevos dígitos => 29
  Último dígito de la suma => 9
        Dígito verificador => 10 - 9 = 1
```

**Desarrolle:**

````python
# Desarrolle solución del EJERCICIO 2 aquí

````

###
**Solución Propuesta:**

````python
def verificar(cedula):
    suma = 0

    # Recorrer los primeros 9 dígitos
    for i in range(9):
        digito = int(cedula[i])

        if i % 2 == 0:      # Si el índice es par
            digito *= 2
            if digito > 9:
                digito -= 9

        suma += digito

    # Último dígito de la suma
    ultimo = suma % 10

    # Calcular el dígito verificador
    if ultimo == 0:
        verificador = 0
    else:
        verificador = 10 - ultimo

    # Comparar con el último dígito de la cédula
    resultado = verificador == int(cedula[-1])
    return resultado
````
